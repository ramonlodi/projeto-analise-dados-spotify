#### ⚠️ Nota metodológica: duplicatas e a seleção de colunas

No trabalho de Estatística foram usadas apenas 10 colunas. Ao restringir o DataFrame a essas colunas, o pandas
reporta linhas repetidas. O bloco abaixo reproduz esse efeito e investiga a causa.

```python
COLUNAS_ESTATISTICA = [
    'track_name', 'track_artist', 'energy', 'tempo', 'danceability',
    'playlist_genre', 'loudness', 'valence', 'track_popularity', 'duration_ms',
]

df_est = df[COLUNAS_ESTATISTICA].copy()

dup_10 = df_est.duplicated().sum()
print(f"Linhas duplicadas usando apenas as 10 colunas: {dup_10}")
print(f"Músicas (track_id) únicas no dataset completo: {df['track_id'].nunique()} de {len(df)}")
print(f"Linhas com track_id repetido:                  {df['track_id'].duplicated().sum()}")
```

"Duplicata" depende do critério adotado. Nos trabalhos originais, o mesmo dataset gerou três números diferentes:

| Critério | Resultado nos trabalhos originais |
|---|---:|
| Linha idêntica nas **29 colunas** (Big Data, Projeto 1) | 0 |
| Linha idêntica nas **10 colunas** selecionadas (Estatística) | 99 |
| Mesmo **`track_id`** (Big Data, Projeto 2 — ML) | 249 (1.686 → 1.437 faixas únicas) |

Não há contradição: são critérios distintos. A mesma faixa pode estar em mais de uma playlist (mudam `playlist_name`,
`playlist_id`, `playlist_subgenre`...), o que é diferente de erro de digitação ou registro repetido. Ao restringir a
10 colunas, descartam-se justamente as colunas que distinguiam as linhas, e surgem repetições. A conclusão correta é:

> *"Não há linhas duplicadas no dataset completo; há faixas repetidas em playlists diferentes."*

**Decisão adotada:** as Partes II e III descrevem as **1.686 entradas** (cada aparição de uma música numa playlist),
como nos trabalhos originais. Já a Parte IV (ML) usa **faixas únicas**, porque a mesma música no treino e no teste
causaria vazamento de dados (seção 13.3). A célula abaixo mede o impacto dessa escolha nas estatísticas descritivas.

```python
# Sensibilidade: estatísticas com e sem repetições (por track_id)
df_unico = df.drop_duplicates(subset='track_id')
cols = ['energy', 'tempo', 'danceability', 'loudness', 'valence', 'track_popularity']

comparacao = pd.DataFrame({
    f'Completo (n={len(df)})': df[cols].mean(),
    f'Sem repetir track_id (n={len(df_unico)})': df_unico[cols].mean(),
})
comparacao['Diferença %'] = (comparacao.iloc[:, 1] / comparacao.iloc[:, 0] - 1) * 100
comparacao.round(4)
```

### 6.3. Conclusão sobre a qualidade dos dados

- ✅ **Sem valores ausentes** (após ler o texto `NA` corretamente).
- ✅ **Sem linhas duplicadas** no dataset completo.
- ℹ️ Algumas músicas aparecem em mais de uma playlist; isso é característica dos dados, não erro.

#### 7.2.2. Gráfico de barras — Modo (maior/menor)

No Spotify, `mode = 1` indica modo **maior** (Major) e `mode = 0` indica modo **menor** (Minor).

```python
# Mapeia explicitamente 0/1 -> rótulo (evita depender da ordem do value_counts)
mapa_modo = {0: 'Minor', 1: 'Major'}
contagem_mode = df['mode'].map(mapa_modo).value_counts()

plt.figure(figsize=(6, 5))
contagem_mode.plot(kind='bar')
plt.title('Distribuição dos Modos Musicais')
plt.xlabel('Modo')
plt.ylabel('Quantidade')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print(contagem_mode)
```

> **Nota de robustez:** a versão original do trabalho atribuía os rótulos com `contagem_mode.index = ['Minor', 'Major']`,
> que rotula **por posição** do `value_counts()` (ordenado do mais para o menos frequente), e não pelo valor de `mode`.
> Isso produz o resultado certo apenas se o modo mais frequente for `0` — o que de fato ocorre neste dataset (≈ 975 Minor
> contra ≈ 710 Major, conforme o gráfico original), então **o resultado publicado estava correto**. Ainda assim, o mapeamento
> explícito `{0: 'Minor', 1: 'Major'}` acima é mais seguro, pois não depende da ordem das frequências.

**Interpretação:** há predominância de músicas em modo *menor*. Na teoria musical, o modo menor costuma ser associado a
emoções mais melancólicas ou introspectivas. O resultado contrasta parcialmente com os altos valores de `energy` e
`danceability`: músicas populares combinam instrumentais energéticos e dançantes com composições emocionalmente mais profundas.

## 8. Análise bivariada

### 8.1. Gráfico de dispersão — Danceability × Popularidade

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='danceability', y='track_popularity', alpha=0.6)
plt.title('Relação entre Danceability e Popularidade')
plt.xlabel('Danceability')
plt.ylabel('Popularidade')
plt.show()

r = df['danceability'].corr(df['track_popularity'])
print(f"Correlação de Pearson (danceability × popularidade): {r:.3f}")
```

## 10. Medidas de tendência central

Representam os valores centrais das distribuições. Serão utilizadas **média**, **mediana** e **moda**.

### 10.1. Média aritmética

Soma de todos os valores dividida pela quantidade de elementos: $\bar{x} = \dfrac{\sum x_i}{n}$

```python
dados_tabela = []
for coluna in colunas_numericas:
    soma = df[coluna].sum()
    quantidade = df[coluna].count()
    dados_tabela.append({
        'Variável': coluna,
        'Soma': soma,
        'Quantidade': quantidade,
        'Média': soma / quantidade,
    })

tabela_medias = pd.DataFrame(dados_tabela)
tabela_medias.style.format({'Soma': '{:,.2f}', 'Quantidade': '{:,}', 'Média': '{:,.2f}'})
```

### 10.2. Mediana

Valor central de um conjunto **ordenado**. Diferente da média, **não é influenciada por valores extremos**, sendo útil
com outliers. Para $n$ par, é a média dos dois valores centrais.

Exemplo passo a passo com a variável `tempo`:

```python
valores = df['tempo'].sort_values().reset_index(drop=True)
n = len(valores)

indice1 = (n // 2) - 1
indice2 = n // 2

recorte = valores.iloc[indice1 - 2 : indice2 + 3]

print('Mediana do atributo Tempo\n')
print(f"{'ID':<10}{'Valor':<15}{'OBS'}")
for i, valor in recorte.items():
    obs = 'Valor Mediano' if i in (indice1, indice2) else '-'
    print(f'{i:<10}{valor:<15.3f}{obs}')

mediana_manual = (valores[indice1] + valores[indice2]) / 2
print(f'\nComo n = {n} é par, mediana = ({valores[indice1]:.3f} + {valores[indice2]:.3f}) / 2 = {mediana_manual:.3f}')
print(f'Conferência com df["tempo"].median(): {df["tempo"].median():.3f}')
```

#### Tabela da mediana de todas as variáveis

```python
tabela_mediana = pd.DataFrame({
    'Variável': colunas_numericas,
    'Mediana': [df[c].median() for c in colunas_numericas],
})
tabela_mediana.style.format({'Mediana': '{:,.2f}'})
```

### 10.3. Moda

Valor que **mais se repete**. Uma variável pode ter mais de uma moda (multimodal).

> **Atenção:** em variáveis **contínuas** (`energy`, `tempo`, `loudness`...), a moda tem pouco significado prático —
> quase todos os valores são distintos e a "moda" é só um valor que por acaso repetiu 5–12 vezes em 1.686.
> Ela é informativa de fato em variáveis **discretas ou categóricas** (`track_popularity`, `playlist_genre`).
> Para as contínuas, a **classe modal** (seção 12.3) é mais adequada.

```python
dados_moda = []
for coluna in colunas_numericas:
    frequencias = df[coluna].value_counts()
    maior_freq = frequencias.max()
    modas = frequencias[frequencias == maior_freq].index.tolist()
    dados_moda.append({
        'Variável': coluna,
        'Moda(s)': ' e '.join(map(str, sorted(modas))),
        'Repetições': maior_freq,
    })

tabela_moda = pd.DataFrame(dados_moda)
print('Tabela de Moda\n')
print(tabela_moda.to_string(index=False))
```

## 11. Medidas de dispersão

**Variância** mede o espalhamento dos dados em relação à média; o **desvio padrão** é sua raiz quadrada (na mesma
unidade da variável, mais fácil de interpretar).

$$s^2 = \frac{\sum (x_i - \bar{x})^2}{n-1} \qquad s = \sqrt{s^2}$$

> O pandas usa por padrão o denominador $n-1$ (**variância amostral**, `ddof=1`). É a convenção adotada aqui.

```python
dispersao = pd.DataFrame({
    'Variável': colunas_numericas,
    'Variância': [df[c].var() for c in colunas_numericas],
    'Desvio Padrão': [df[c].std() for c in colunas_numericas],
})
dispersao['Coef. de Variação (%)'] = dispersao['Desvio Padrão'] / df[colunas_numericas].mean().abs().values * 100
dispersao.round(4)
```

**Como ler:** variâncias de variáveis com **escalas diferentes não são comparáveis diretamente** (a de `duration_ms` é
da ordem de 10⁹ só porque a unidade é milissegundo). O **coeficiente de variação** (desvio ÷ |média|) padroniza a
comparação: permite ver, por exemplo, que `valence` e `energy` variam relativamente muito mais que `track_popularity`,
que é a mais concentrada — coerente com o fato de o dataset conter só músicas já populares.

## 12. Dados agrupados em classes

Com dados contínuos como `tempo`, é comum agrupá-los em **classes** (intervalos) e calcular as medidas a partir do
**ponto médio** e da **frequência** de cada classe. Usa-se aqui `pd.cut` com 5 classes de mesma amplitude.

### 12.1. Distribuição de frequências

```python
N_CLASSES = 5
classes = pd.cut(df['tempo'], bins=N_CLASSES)

freq = classes.value_counts().sort_index()

tab = pd.DataFrame({
    'Classe': freq.index.astype(str),
    'Limite inf.': [iv.left for iv in freq.index],
    'Limite sup.': [iv.right for iv in freq.index],
    'Ponto Médio': [iv.mid for iv in freq.index],
    'Quantidade': freq.values,
})
tab['Porcentagem'] = tab['Quantidade'] / tab['Quantidade'].sum() * 100
tab['Freq. Acumulada'] = tab['Quantidade'].cumsum()
tab
```

> `pd.cut` estende ligeiramente o limite inferior da 1ª classe para incluir o valor mínimo; por isso os limites
> impressos (ex.: 49,145) ficam um pouco abaixo do mínimo real (49,305).

### 12.2. Média agrupada

$$\bar{x}_{agr} = \frac{\sum (PM_i \cdot f_i)}{\sum f_i}$$

```python
n_total = tab['Quantidade'].sum()
media_agrupada = (tab['Ponto Médio'] * tab['Quantidade']).sum() / n_total

print(f'Média agrupada:              {media_agrupada:.5f}')
print(f'Média real (dados brutos):   {df["tempo"].mean():.5f}')
print(f'Diferença (erro de agrupamento): {media_agrupada - df["tempo"].mean():+.5f}')
```

### 12.3. Mediana agrupada e classe modal

**Mediana agrupada** (por interpolação), onde $L$ = limite inferior da classe mediana, $F_{ant}$ = frequência
acumulada anterior, $f$ = frequência da classe mediana e $h$ = amplitude da classe:

$$Md = L + \left(\frac{n/2 - F_{ant}}{f}\right) \cdot h$$

**Moda de Czuber**, onde $d_1$ e $d_2$ são as diferenças entre a frequência da classe modal e as das classes
anterior e posterior:

$$Mo = L + \left(\frac{d_1}{d_1 + d_2}\right) \cdot h$$

```python
h = tab['Limite sup.'].iloc[0] - tab['Limite inf.'].iloc[0]   # amplitude da classe

# ---- Mediana agrupada
i_med = (tab['Freq. Acumulada'] >= n_total / 2).idxmax()
L = tab.loc[i_med, 'Limite inf.']
F_ant = tab.loc[i_med - 1, 'Freq. Acumulada'] if i_med > 0 else 0
f_med = tab.loc[i_med, 'Quantidade']
mediana_agrupada = L + ((n_total / 2 - F_ant) / f_med) * h

# ---- Classe modal e moda de Czuber
i_mod = tab['Quantidade'].idxmax()
f_mod = tab.loc[i_mod, 'Quantidade']
f_ant = tab.loc[i_mod - 1, 'Quantidade'] if i_mod > 0 else 0
f_pos = tab.loc[i_mod + 1, 'Quantidade'] if i_mod < len(tab) - 1 else 0
d1, d2 = f_mod - f_ant, f_mod - f_pos
moda_czuber = tab.loc[i_mod, 'Limite inf.'] + d1 / (d1 + d2) * h

print(f'Classe mediana: {tab.loc[i_med, "Classe"]}')
print(f'Mediana agrupada: {mediana_agrupada:.3f}   (mediana real: {df["tempo"].median():.3f})')
print()
print(f'Classe modal:    {tab.loc[i_mod, "Classe"]}  ({f_mod} músicas, {tab.loc[i_mod, "Porcentagem"]:.2f}%)')
print(f'Moda de Czuber:  {moda_czuber:.3f}')
```

### 12.4. Variância e desvio padrão agrupados

Desvio quadrático de cada classe: $(PM_i - \bar{x}_{agr})^2$, ponderado pela frequência.

$$s^2_{agr} = \frac{\sum f_i (PM_i - \bar{x}_{agr})^2}{n - 1}$$

```python
tab['Desvio Quadrático'] = (tab['Ponto Médio'] - media_agrupada) ** 2
tab['Desvio Quad. × Freq.'] = tab['Desvio Quadrático'] * tab['Quantidade']

soma_dq = tab['Desvio Quad. × Freq.'].sum()
var_agrupada = soma_dq / (n_total - 1)
dp_agrupado = var_agrupada ** 0.5

display(tab[['Classe', 'Ponto Médio', 'Quantidade', 'Porcentagem', 'Desvio Quadrático', 'Desvio Quad. × Freq.']].round(6))

print(f'\nΣ f·(PM − x̄)² = {soma_dq:,.2f}')
print(f'Variância agrupada:      {var_agrupada:.2f}   | dados brutos: {df["tempo"].var():.2f}')
print(f'Desvio padrão agrupado:  {dp_agrupado:.2f}    | dados brutos: {df["tempo"].std():.2f}')
```

**Interpretação:** o agrupamento é uma **aproximação**, pois assume que todos os valores de uma classe são iguais ao
ponto médio, descartando a variação *dentro* de cada classe. Compare os valores agrupados com os obtidos nos dados
brutos (impressos acima): média e mediana ficam muito próximas, enquanto a variância agrupada difere mais.

> Com os dados originais do trabalho, a média agrupada foi 121,52 (real 121,07) e a variância agrupada ≈ 801 (real ≈ 733),
> ou seja, **≈ 9% maior**. Uma correção clássica para esse efeito é a *correção de Sheppard*
> ($s^2_{corrigida} \approx s^2_{agr} - h^2/12$), que aproxima melhor o valor real. O resultado exato depende do número
> de classes escolhido: mais classes → aproximação melhor.

```python
# Histograma com as classes usadas
plt.figure(figsize=(9, 5))
plt.bar(tab['Ponto Médio'], tab['Quantidade'], width=h * 0.95, edgecolor='black', alpha=0.8)
plt.axvline(media_agrupada, color='red', ls='--', label=f'Média agrupada ({media_agrupada:.1f})')
plt.axvline(mediana_agrupada, color='green', ls='--', label=f'Mediana agrupada ({mediana_agrupada:.1f})')
plt.axvline(moda_czuber, color='purple', ls='--', label=f'Moda de Czuber ({moda_czuber:.1f})')
plt.title('Distribuição de Tempo (BPM) em 5 classes')
plt.xlabel('Tempo (BPM)')
plt.ylabel('Quantidade de músicas')
plt.legend()
plt.tight_layout()
plt.show()
```

### 13.1. Objetivo e conjunto de dados desta parte

Aplicar **aprendizado supervisionado (classificação multiclasse)** para prever o gênero de uma música, comparando
**quatro modelos** (Regressão Logística, KNN, Árvore de Decisão e Random Forest) e tomando cuidado para evitar

Atenção: esta parte **não usa exatamente as mesmas linhas** das Partes II e III, e isso é intencional:

| Partes | Base utilizada | Linhas |
|---|---|---:|
| II e III (exploratória e estatística) | Dataset completo — cada linha é uma música **em uma playlist** | 1.686 |

> **Interpretação do alvo:** `playlist_genre` é o gênero da **playlist** em que a música aparece, e não um rótulo

Foram mantidas as variáveis sonoras e numéricas já analisadas nas partes anteriores. Foram descartados
identificadores, links, nomes (música, artista, álbum), data de lançamento e `playlist_name`.

**`playlist_subgenre` foi propositalmente descartada**: ela está diretamente relacionada ao gênero (o subgênero

A coluna `track_id` é mantida **temporariamente**, apenas para remover repetições na próxima etapa.

### 13.3. Limpeza: faixas repetidas

A mesma faixa pode estar em várias playlists (linhas diferentes, mesmo `track_id`). O diagnóstico abaixo mede
quantas linhas são repetições e, principalmente, **quantas faixas aparecem em playlists de gêneros diferentes**.

```python
n_linhas = len(df_ml)
n_unicas = df_ml['track_id'].nunique()
n_repetidas = df_ml['track_id'].duplicated().sum()

print(f'Linhas:                                   {n_linhas}')
print(f'Faixas únicas (track_id):                 {n_unicas}')
print(f'Linhas com track_id já visto antes:       {n_repetidas}')

# Faixas presentes em playlists de gêneros diferentes
generos_por_faixa = df_ml.groupby('track_id')[ALVO].nunique()
n_multi = int((generos_por_faixa > 1).sum())
print(f'\nFaixas associadas a mais de um gênero:    {n_multi}')
print(f'Faixas repetidas com o MESMO gênero:      {int(((df_ml.groupby("track_id").size() > 1) & (generos_por_faixa == 1)).sum())}')
```

No trabalho original, esse diagnóstico encontrou **249 linhas repetidas** (1.686 → **1.437** faixas únicas).
Isso também explica a divergência entre os trabalhos: o de Estatística encontrou 99 linhas idênticas nas 10 colunas
que usou (seção 6.2), enquanto "mesmo `track_id`" é um critério mais amplo.

A decisão foi manter **uma linha por faixa** (`keep='first'`). Duas consequências devem ser registradas:

1. **Evita vazamento:** a mesma música não aparece mais em treino e teste ao mesmo tempo.
2. **Introduz ruído de rótulo:** para faixas presentes em playlists de gêneros diferentes, o gênero mantido é o da
   primeira ocorrência no arquivo, o que é uma escolha arbitrária. O número de faixas nessa situação (célula

```python
df_ml = df_ml.drop_duplicates(subset='track_id', keep='first').reset_index(drop=True)
df_ml = df_ml.drop(columns='track_id')

print('Formato após remover repetições:', df_ml.shape)
print('Valores nulos nas colunas utilizadas:', int(df_ml.isnull().sum().sum()))
```

```python
plt.figure(figsize=(10, 8))
cols_num_ml = df_ml.select_dtypes(include=[np.number]).columns
sns.heatmap(df_ml[cols_num_ml].corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Matriz de Correlação (dataset de ML)')
plt.tight_layout()
plt.show()
```

A matriz confirma o que a seção 9 já mostrava: `energy` e `loudness` são fortemente correlacionadas (≈ 0,69) e
`acousticness` se opõe a `energy` (≈ −0,60). Variáveis correlacionadas **dividem** a importância entre si nos
modelos, o que será considerado ao interpretar a seção 13.8.5.

### 13.6. Preparação dos dados

Cuidados para evitar vazamento de dados:

- **Separar treino e teste antes de qualquer padronização;**
- Ajustar o `StandardScaler` (`fit`) **somente com o treino** e apenas aplicar (`transform`) ao teste;
- Usar `stratify=y` para manter a proporção dos gêneros nos dois conjuntos (80% treino / 20% teste).

A padronização é necessária para Regressão Logística e KNN (sensíveis à escala); as árvores não precisam dela, mas
não são prejudicadas.

```python
X = df_ml.drop(columns=[ALVO])
y = df_ml[ALVO]

print('Variáveis preditoras (X):', X.columns.tolist())
print('Variável alvo (y):', ALVO)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)
print('\nTreino:', X_train.shape, '| Teste:', X_test.shape)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)   # ajusta SOMENTE no treino
X_test_scaled = scaler.transform(X_test)         # apenas aplica no teste
```

### 13.7. Modelagem

Quatro modelos, do mais simples ao mais complexo, mais uma **referência de comparação**:

0. **Baseline** (`DummyClassifier`): prevê sempre o gênero mais frequente. Qualquer modelo útil precisa superá-lo;
1. **Regressão Logística**: modelo linear;
2. **KNN** (k = 5): baseado na similaridade entre músicas;
3. **Árvore de Decisão** (profundidade máxima 10): regras simples;
4. **Random Forest** (200 árvores): conjunto de árvores.

Não foi feito ajuste de hiperparâmetros: os valores são padrões razoáveis, e não o resultado de uma busca.

```python
# Referência: sempre prever o gênero mais frequente
baseline = DummyClassifier(strategy='most_frequent').fit(X_train_scaled, y_train)
acc_baseline = accuracy_score(y_test, baseline.predict(X_test_scaled))

modelos = {
    'Regressão Logística': LogisticRegression(max_iter=1000, random_state=RANDOM_STATE),
    'KNN': KNeighborsClassifier(n_neighbors=5),
    'Árvore de Decisão': DecisionTreeClassifier(random_state=RANDOM_STATE, max_depth=10),
    'Random Forest': RandomForestClassifier(n_estimators=200, random_state=RANDOM_STATE),
}

resultados = []
predicoes = {}

for nome, modelo in modelos.items():
    modelo.fit(X_train_scaled, y_train)
    pred = modelo.predict(X_test_scaled)
    predicoes[nome] = pred

    acuracia = accuracy_score(y_test, pred)
    f1 = f1_score(y_test, pred, average='weighted')
    resultados.append({'Modelo': nome, 'Acurácia': acuracia, 'F1-Score': f1})
    print(f'{nome} treinado. Acurácia: {acuracia:.3f} | F1-Score: {f1:.3f}')

genero_baseline = y_train.value_counts().idxmax()
print(f'\nBaseline (sempre "{genero_baseline}"): {acc_baseline:.3f}')
print(f'Sorteio uniforme entre {y.nunique()} classes (referência teórica): {1 / y.nunique():.3f}')
```

Os comentários desta seção se referem à **execução original** (Kaggle), na qual o Random Forest foi o melhor
modelo. Reexecutar com outra versão do scikit-learn pode alterar levemente os números.

#### 13.8.1. Comparação geral

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=resultados_df, x='Modelo', y='Acurácia')
plt.axhline(acc_baseline, color='red', ls='--', label=f'Baseline: sempre o gênero mais frequente ({acc_baseline:.3f})')
plt.axhline(1 / y.nunique(), color='gray', ls=':', label=f'Sorteio uniforme ({1 / y.nunique():.2f})')
plt.title('Comparação de Acurácia entre os Modelos')
plt.ylim(0, 1)
plt.xticks(rotation=20)
plt.legend(loc='upper right', fontsize=8)
plt.tight_layout()
plt.show()
```

**Resultados da execução original (253 músicas de teste):**

| Modelo | Acurácia | F1 (ponderado) | Acertos |
|---|---:|---:|---:|
| Random Forest | 0,455 | 0,436 | 115 |
| Regressão Logística | 0,451 | 0,433 | 114 |
| KNN | 0,399 | 0,395 | 101 |
| Árvore de Decisão | 0,336 | 0,335 | 85 |

**Como ler esses números**

- **Contra o acaso:** um sorteio uniforme entre 10 gêneros acertaria ≈ 10%. Mas a referência mais justa é o
  **baseline "sempre pop"**: como o *pop* é 60 das 253 músicas de teste, ele acerta **≈ 23,7%**. Os modelos superam esse
  patamar (o melhor chega a ≈ 45%), então as características sonoras têm poder preditivo real, porém moderado.
- **Random Forest × Regressão Logística:** a diferença é de **um único acerto** (115 contra 114 em 253). Com um só
  conjunto de teste pequeno, isso deve ser tratado como **empate técnico**, e não como superioridade do Random Forest.
  A validação cruzada abaixo dá uma estimativa mais estável.
- **Árvore de Decisão:** a árvore única é claramente pior, o que é esperado: ela tende a sobreajustar (*overfitting*)
  em conjuntos pequenos, e o Random Forest reduz isso ao combinar muitas árvores.

#### 13.8.2. Validação cruzada (complemento)

Um único *split* de 253 exemplos é sensível ao acaso da divisão. A validação cruzada estratificada de 5 partes
avalia cada modelo em 5 divisões diferentes. O `StandardScaler` fica **dentro do pipeline**, sendo reajustado a cada
partição, para que o teste de cada partição nunca influencie a padronização.

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

linhas_cv = []
candidatos = {'Baseline (mais frequente)': DummyClassifier(strategy='most_frequent'), **modelos}

for nome, modelo in candidatos.items():
    pipe = make_pipeline(StandardScaler(), clone(modelo))
    scores = cross_val_score(pipe, X, y, cv=cv, scoring='accuracy')
    linhas_cv.append({'Modelo': nome, 'Acurácia média': scores.mean(), 'Desvio padrão': scores.std()})

pd.DataFrame(linhas_cv).sort_values('Acurácia média', ascending=False).round(3).reset_index(drop=True)
```

Interprete a **média ± desvio padrão**: se dois modelos têm médias próximas e o desvio é comparável à diferença
entre elas, não há evidência de que um seja melhor que o outro.

## 14. Conclusão geral

O conjunto das análises permite um retrato consistente das músicas populares do Spotify:

**Perfil e distribuição (Partes II e III)**

1. **Perfil predominante:** músicas energéticas, dançantes, com volume elevado, ~120 BPM e ~3,5 minutos de duração — características associadas à música comercial contemporânea.
2. **Concentração de gêneros:** o pop domina, seguido por rock, hip-hop, latin e electronic, com uma longa cauda de gêneros de nicho.
3. **Diferenças entre gêneros:** metal, punk, k-pop e electronic são os mais energéticos; classical, jazz, lofi e folk, os menos.
4. **Relações entre variáveis:** energy, loudness e danceability estão relacionadas, e acousticness se opõe a energy e loudness.
5. **Estatística:** as medidas de centralidade e dispersão mostram distribuições relativamente equilibradas; o tratamento de dados agrupados reproduz bem a média e a mediana, com aproximação menos precisa para a variância.

**Popularidade × gênero (o contraste central do trabalho)**

6. **Popularidade:** tem baixa correlação com os atributos sonoros e baixa dispersão (desvio ≈ 6). Entre músicas já populares, o som **quase não explica** quem é mais popular — provavelmente há influência de fatores externos (algoritmos, marketing, viralização).
7. **Gênero:** ao contrário da popularidade, o gênero **é parcialmente previsível** pelo som (≈ 45% de acerto contra ≈ 24% do baseline), embora longe de ser confiável.

Os dois resultados se complementam: as características sonoras descrevem **o estilo** de uma música razoavelmente bem,
mas descrevem **o seu sucesso** muito mal. E o limite na previsão do gênero é coerente com a alta dispersão interna
dos gêneros vista na Parte II.

### Limitações

- O dataset contém **apenas músicas já populares**, o que restringe a amplitude de `track_popularity` e enfraquece correlações. Não se pode concluir que o som seja irrelevante para a popularidade em geral.
- Correlação não implica causalidade.
- O alvo é o gênero da **playlist**, e não da faixa; faixas presentes em várias playlists foram reduzidas a uma linha (`keep='first'`), o que introduz ruído de rótulo.
- O modelo de ML cobre só os **10 gêneros com ≥ 30 faixas**; sem tratamento de desbalanceamento e sem ajuste de hiperparâmetros.
- Avaliação principal em um único *split* de 253 músicas (complementada por validação cruzada).
- O corte de "popular" e a data de extração do dataset afetam o retrato obtido.
